## Bangla Digit Normalization

Convert Bengali digits to ASCII digits before matching land identifiers.

```python
BANGLA_DIGITS = str.maketrans(
    "০১২৩৪৫৬৭৮৯",
    "0123456789"
)

def bangla_to_english_digits(value):
    if value is None:
        return None

    value = str(value)
    return value.translate(BANGLA_DIGITS)
```

```python
print(bangla_to_english_digits("৭৩১"))
print(bangla_to_english_digits("২০৫"))
print(bangla_to_english_digits("০০৭৩১"))
```

```python
def normalize_land_number(value):
    if value is None:
        return None

    value = bangla_to_english_digits(value)

    value = unicodedata.normalize("NFKC", value)

    value = value.strip()

    labels = [
        "দাগ নং",
        "দাগ নম্বর",
        "দাগ",
        "খতিয়ান নং",
        "খতিয়ান নং",
        "খতিয়ান নম্বর",
        "খতিয়ান নম্বর",
        "খতিয়ান",
        "খতিয়ান",
        "নং",
        "নম্বর",
        "NO.",
        "NO"
    ]

    for label in labels:
        value = value.replace(label, "")

    value = value.strip()

    # Current MVP: retain numeric characters only
    digits = re.sub(r"[^0-9]", "", value)

    if digits == "":
        return None

    # Remove unnecessary leading zeros
    return str(int(digits))
```

```python
test_values = [
    "731",
    "৭৩১",
    "০০৭৩১",
    "দাগ নং ৭৩১",
    "দাগ নম্বর: 731",
]

for value in test_values:
    print(
        repr(value),
        "->",
        normalize_land_number(value)
    )
```

```python
SURVEY_ALIASES = {
    "CS": "CS",
    "C S": "CS",
    "C.S": "CS",
    "C.S.": "CS",
    "সি এস": "CS",

    "SA": "SA",
    "S A": "SA",
    "S.A": "SA",
    "S.A.": "SA",
    "এস এ": "SA",

    "RS": "RS",
    "R S": "RS",
    "R.S": "RS",
    "R.S.": "RS",
    "আর এস": "RS",

    "BS": "BS",
    "B S": "BS",
    "B.S": "BS",
    "B.S.": "BS",
    "বি এস": "BS",
}

def normalize_survey_type(value):
    if value is None:
        return None

    value = unicodedata.normalize(
        "NFKC",
        str(value)
    )

    value = " ".join(value.strip().upper().split())

    return SURVEY_ALIASES.get(value, value)
```

```python
survey_tests = [
    "RS",
    "R.S.",
    "r s",
    "আর এস",
    "C.S.",
    "সি এস",
    "বি এস"
]

for value in survey_tests:
    print(
        repr(value),
        "->",
        normalize_survey_type(value)
    )
```

```python
def normalize_location(value):
    if value is None:
        return None

    value = unicodedata.normalize(
        "NFKC",
        str(value)
    )

    value = " ".join(value.strip().split())

    return value.upper()
```

```python
location_tests = [
    "Demo-District-A",
    " demo-district-a ",
    "Demo-District-A   "
]

for value in location_tests:
    print(
        repr(value),
        "->",
        normalize_location(value)
    )
```

```python
def create_parcel_key(
    district,
    upazila,
    mouza,
    survey_type,
    khatian_no,
    dag_no
):
    district_n = normalize_location(district)
    upazila_n = normalize_location(upazila)
    mouza_n = normalize_location(mouza)

    survey_n = normalize_survey_type(survey_type)

    khatian_n = normalize_land_number(khatian_no)
    dag_n = normalize_land_number(dag_no)

    components = [
        district_n,
        upazila_n,
        mouza_n,
        survey_n,
        khatian_n,
        dag_n
    ]

    if any(value is None for value in components):
        return None

    return "|".join(components)
```

```python
normalized_parcels_df = parcels_df.copy()

normalized_parcels_df["parcel_key"] = normalized_parcels_df.apply(
    lambda row: create_parcel_key(
        row["district"],
        row["upazila"],
        row["mouza"],
        row["survey_type"],
        row["khatian_no"],
        row["dag_no"]
    ),
    axis=1
)

normalized_parcels_df[
    [
        "parcel_id",
        "parcel_key"
    ]
]
```

```python
def get_owner_area(ledger_df, parcel_id, owner_id):
    rows = ledger_df[
        (ledger_df["parcel_id"] == parcel_id) &
        (ledger_df["owner_id"] == owner_id) &
        (ledger_df["status"] == "ACTIVE")
    ]

    if rows.empty:
        return 0.0

    return float(rows["owned_area_decimal"].sum())
```

```python
def next_ownership_id(ledger_df):
    numbers = []

    for value in ledger_df["ownership_id"]:
        value = str(value)

        if value.startswith("OWN"):
            number = value.replace("OWN", "")

            if number.isdigit():
                numbers.append(int(number))

    next_number = max(numbers, default=0) + 1

    return f"OWN{next_number:03d}"
```

```python
print(
    "Next ownership ID:",
    next_ownership_id(ownership_df)
)
```

```python
def check_ledger_integrity(ledger_df, parcels_df):

    active = ledger_df[
        ledger_df["status"] == "ACTIVE"
    ].copy()

    totals = (
        active
        .groupby("parcel_id")["owned_area_decimal"]
        .sum()
        .reset_index()
    )

    totals = totals.merge(
        parcels_df[
            ["parcel_id", "total_area_decimal"]
        ],
        on="parcel_id",
        how="left"
    )

    totals["valid"] = (
        (totals["owned_area_decimal"] >= 0) &
        (
            totals["owned_area_decimal"]
            <= totals["total_area_decimal"]
        )
    )

    negative_rows = ledger_df[
        ledger_df["owned_area_decimal"] < 0
    ]

    return (
        totals["valid"].all()
        and negative_rows.empty
    )
```

```python
def process_transfer(
    ledger_df,
    transaction,
    people_df,
    parcels_df
):
    # Work on a copy
    updated_ledger = ledger_df.copy()

    seller_id = transaction["seller_id"]
    buyer_id = transaction["buyer_id"]
    parcel_id = transaction["parcel_id"]

    transfer_area = float(
        transaction["transferred_area_decimal"]
    )

    # Basic reference checks
    person_ids = set(people_df["person_id"])
    parcel_ids = set(parcels_df["parcel_id"])

    if seller_id not in person_ids:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "SELLER_NOT_FOUND"
        }

    if buyer_id not in person_ids:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "BUYER_NOT_FOUND"
        }

    if parcel_id not in parcel_ids:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "PARCEL_NOT_FOUND"
        }

    if seller_id == buyer_id:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "SELLER_BUYER_SAME"
        }

    if transfer_area <= 0:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "INVALID_TRANSFER_AREA"
        }

    # Seller ownership check
    seller_available = get_owner_area(
        updated_ledger,
        parcel_id,
        seller_id
    )

    if seller_available == 0:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "SELLER_NOT_OWNER",
            "seller_available": seller_available
        }

    if transfer_area > seller_available:
        return ledger_df.copy(), {
            "valid": False,
            "reason": "INSUFFICIENT_OWNERSHIP",
            "seller_available": seller_available,
            "requested_area": transfer_area
        }

    # Update seller
    seller_mask = (
        (updated_ledger["parcel_id"] == parcel_id) &
        (updated_ledger["owner_id"] == seller_id) &
        (updated_ledger["status"] == "ACTIVE")
    )

    seller_index = updated_ledger[
        seller_mask
    ].index[0]

    updated_ledger.loc[
        seller_index,
        "owned_area_decimal"
    ] -= transfer_area

    # Seller owns nothing after full transfer
    if (
        updated_ledger.loc[
            seller_index,
            "owned_area_decimal"
        ] == 0
    ):
        updated_ledger.loc[
            seller_index,
            "status"
        ] = "INACTIVE"

    # Check whether buyer already owns part of parcel
    buyer_mask = (
        (updated_ledger["parcel_id"] == parcel_id) &
        (updated_ledger["owner_id"] == buyer_id) &
        (updated_ledger["status"] == "ACTIVE")
    )

    if buyer_mask.any():

        buyer_index = updated_ledger[
            buyer_mask
        ].index[0]

        updated_ledger.loc[
            buyer_index,
            "owned_area_decimal"
        ] += transfer_area

    else:

        new_row = {
            "ownership_id":
                next_ownership_id(updated_ledger),

            "parcel_id":
                parcel_id,

            "owner_id":
                buyer_id,

            "owned_area_decimal":
                transfer_area,

            "valid_from":
                transaction["execution_date"],

            "status":
                "ACTIVE"
        }

        updated_ledger = pd.concat(
            [
                updated_ledger,
                pd.DataFrame([new_row])
            ],
            ignore_index=True
        )

    # Final safety check
    if not check_ledger_integrity(
        updated_ledger,
        parcels_df
    ):
        return ledger_df.copy(), {
            "valid": False,
            "reason": "LEDGER_INTEGRITY_FAILURE"
        }

    return updated_ledger, {
        "valid": True,
        "reason": "VALID_TRANSFER",
        "seller_available_before":
            seller_available,
        "transferred_area":
            transfer_area
    }
```


# STEP 11 — Double-Selling Detection

Detect whether the same identifiable land portion has already been
transferred in an earlier valid transaction.

For the synthetic MVP, each transferable portion may have an internal
`portion_id`.

Scenario:

- PAR001 total area = 20 decimal
- DEED001 already sold portion PAR001-PART-A (5 decimal) to P002
- P001 still owns another 15 decimal
- P001 attempts to sell PAR001-PART-A again to P003

The area check alone would not catch this because P001 still owns 15 decimal.

Expected result:

- DOUBLE_SALE detected
- transaction rejected
- ownership ledger unchanged
- transaction history unchanged


```python
accepted_transactions_df = transactions_df.copy()

accepted_transactions_df["portion_id"] = [
    "PAR001-PART-A"
]

accepted_transactions_df[
    [
        "deed_id",
        "seller_id",
        "buyer_id",
        "parcel_id",
        "transferred_area_decimal",
        "portion_id",
        "transaction_status"
    ]
]
```

```python
def check_double_sale(history_df, transaction):

    portion_id = transaction.get("portion_id")

    if portion_id is None:
        return {
            "double_sale": False,
            "reason": "NO_PORTION_ID"
        }

    previous_sales = history_df[
        (history_df["parcel_id"] == transaction["parcel_id"]) &
        (history_df["portion_id"] == portion_id) &
        (history_df["transaction_status"] == "VALID")
    ]

    if previous_sales.empty:
        return {
            "double_sale": False,
            "reason": "NO_PREVIOUS_SALE"
        }

    previous_deeds = previous_sales[
        "deed_id"
    ].tolist()

    previous_buyers = previous_sales[
        "buyer_id"
    ].tolist()

    return {
        "double_sale": True,
        "reason": "DOUBLE_SALE",
        "previous_deeds": previous_deeds,
        "previous_buyers": previous_buyers
    }
```

```python
def process_transfer_with_history(
    ledger_df,
    history_df,
    transaction,
    people_df,
    parcels_df
):

    # 1. Check historical portion conflict first
    double_sale_result = check_double_sale(
        history_df,
        transaction
    )

    if double_sale_result["double_sale"]:

        return (
            ledger_df.copy(),
            history_df.copy(),
            {
                "valid": False,
                "reason": "DOUBLE_SALE",
                "previous_deeds":
                    double_sale_result["previous_deeds"],
                "previous_buyers":
                    double_sale_result["previous_buyers"]
            }
        )

    # 2. Run normal ownership validation
    updated_ledger, transfer_result = process_transfer(
        ledger_df,
        transaction,
        people_df,
        parcels_df
    )

    if not transfer_result["valid"]:

        return (
            ledger_df.copy(),
            history_df.copy(),
            transfer_result
        )

    # 3. Add accepted transaction to history
    accepted_transaction = transaction.copy()

    accepted_transaction[
        "transaction_status"
    ] = "VALID"

    updated_history = pd.concat(
        [
            history_df,
            pd.DataFrame(
                [accepted_transaction]
            )
        ],
        ignore_index=True
    )

    return (
        updated_ledger,
        updated_history,
        {
            "valid": True,
            "reason": "VALID_TRANSFER"
        }
    )
```

# STEP 12 — Duplicate Deed Number Detection

Detect whether a deed identifier has already been used in the accepted
transaction history.

Expected behavior:

- Duplicate deed ID is rejected before ownership is modified
- Ledger remains unchanged
- Transaction history remains unchanged

```python
def check_duplicate_deed(history_df, deed_id):

    matches = history_df[
        history_df["deed_id"] == deed_id
    ]

    if matches.empty:
        return {
            "duplicate": False,
            "reason": "DEED_ID_AVAILABLE"
        }

    return {
        "duplicate": True,
        "reason": "DUPLICATE_DEED",
        "existing_records": len(matches)
    }
```

```python
def check_deed_mutation_match(
    mutation_df,
    transaction
):

    records = mutation_df[
        (mutation_df["parcel_id"] ==
         transaction["parcel_id"]) &
        (mutation_df["portion_id"] ==
         transaction["portion_id"]) &
        (mutation_df["status"] == "ACTIVE")
    ]

    if records.empty:
        return {
            "match": None,
            "reason": "NO_MUTATION_RECORD"
        }

    mutation_owners = set(
        records["mutation_owner_id"]
    )

    deed_seller = transaction["seller_id"]

    if deed_seller in mutation_owners:
        return {
            "match": True,
            "reason": "MUTATION_MATCH",
            "mutation_owners":
                sorted(mutation_owners)
        }

    return {
        "match": False,
        "reason": "DEED_MUTATION_MISMATCH",
        "deed_seller": deed_seller,
        "mutation_owners":
            sorted(mutation_owners)
    }
```

# STEP 14 — Unified Explainable Risk Engine

Combine all fraud/conflict checks into one explainable verification report.

Risk policy for the MVP:

- LOW: all available checks pass
- MEDIUM: no critical conflict, but some information is unavailable
- HIGH: one or more critical conflicts are detected

Important:
Risk level is a screening result, not a legal declaration of fraud or ownership.

```python
def evaluate_transaction_rules(
    ledger_df,
    history_df,
    mutation_df,
    transaction,
    people_df,
    parcels_df
):
    rules = {}

    seller_id = transaction["seller_id"]
    buyer_id = transaction["buyer_id"]
    parcel_id = transaction["parcel_id"]
    deed_id = transaction["deed_id"]

    transfer_area = float(
        transaction["transferred_area_decimal"]
    )

    # ---------------------------------
    # Rule 1: Deed ID
    # ---------------------------------
    duplicate_result = check_duplicate_deed(
        history_df,
        deed_id
    )

    if duplicate_result["duplicate"]:
        rules["deed_id_check"] = {
            "status": "FAIL",
            "code": "DUPLICATE_DEED",
            "message": (
                f"Deed ID {deed_id} already exists."
            )
        }
    else:
        rules["deed_id_check"] = {
            "status": "PASS",
            "code": "DEED_ID_AVAILABLE",
            "message": "Deed ID is not duplicated."
        }

    # ---------------------------------
    # Rule 2: Parcel
    # ---------------------------------
    parcel_exists = (
        parcel_id in set(parcels_df["parcel_id"])
    )

    if parcel_exists:
        rules["parcel_check"] = {
            "status": "PASS",
            "code": "PARCEL_FOUND",
            "message": "Parcel exists in the registry."
        }
    else:
        rules["parcel_check"] = {
            "status": "FAIL",
            "code": "PARCEL_NOT_FOUND",
            "message": "Parcel was not found."
        }

    # ---------------------------------
    # Rule 3: Seller ownership
    # ---------------------------------
    seller_area = get_owner_area(
        ledger_df,
        parcel_id,
        seller_id
    )

    if seller_area > 0:
        rules["seller_ownership_check"] = {
            "status": "PASS",
            "code": "SELLER_IS_OWNER",
            "message": (
                f"Seller currently owns "
                f"{seller_area} decimal."
            )
        }
    else:
        rules["seller_ownership_check"] = {
            "status": "FAIL",
            "code": "SELLER_NOT_OWNER",
            "message": (
                "Seller has no active ownership "
                "in this parcel."
            )
        }

    # ---------------------------------
    # Rule 4: Transfer area
    # ---------------------------------
    if transfer_area <= 0:
        rules["area_check"] = {
            "status": "FAIL",
            "code": "INVALID_TRANSFER_AREA",
            "message": "Transfer area must be positive."
        }

    elif transfer_area > seller_area:
        rules["area_check"] = {
            "status": "FAIL",
            "code": "INSUFFICIENT_OWNERSHIP",
            "message": (
                f"Requested {transfer_area} decimal, "
                f"but seller has {seller_area} decimal."
            )
        }

    else:
        rules["area_check"] = {
            "status": "PASS",
            "code": "AREA_AVAILABLE",
            "message": (
                f"Requested {transfer_area} decimal "
                "is within seller ownership."
            )
        }

    # ---------------------------------
    # Rule 5: Double-sale
    # ---------------------------------
    double_sale_result = check_double_sale(
        history_df,
        transaction
    )

    if double_sale_result["double_sale"]:
        rules["double_sale_check"] = {
            "status": "FAIL",
            "code": "DOUBLE_SALE",
            "message": (
                "The same identified land portion "
                "appears in a previous valid sale."
            )
        }

    elif double_sale_result["reason"] == "NO_PORTION_ID":
        rules["double_sale_check"] = {
            "status": "WARNING",
            "code": "PORTION_ID_UNAVAILABLE",
            "message": (
                "Double-sale verification is incomplete "
                "because no portion ID is available."
            )
        }

    else:
        rules["double_sale_check"] = {
            "status": "PASS",
            "code": "NO_PREVIOUS_SALE",
            "message": (
                "No previous valid sale was found "
                "for this portion."
            )
        }

    # ---------------------------------
    # Rule 6: Mutation consistency
    # ---------------------------------
    mutation_result = check_deed_mutation_match(
        mutation_df,
        transaction
    )

    if mutation_result["match"] is True:
        rules["mutation_check"] = {
            "status": "PASS",
            "code": "MUTATION_MATCH",
            "message": (
                "Deed seller matches the active "
                "mutation-side owner."
            )
        }

    elif mutation_result["match"] is False:
        rules["mutation_check"] = {
            "status": "FAIL",
            "code": "DEED_MUTATION_MISMATCH",
            "message": (
                f"Deed seller is {seller_id}, "
                f"but mutation record lists "
                f"{', '.join(mutation_result['mutation_owners'])}."
            )
        }

    else:
        rules["mutation_check"] = {
            "status": "WARNING",
            "code": "NO_MUTATION_RECORD",
            "message": (
                "No matching mutation record is available "
                "for this parcel portion."
            )
        }

    return rules
```

```python
def calculate_risk_level(rules):

    fail_count = sum(
        rule["status"] == "FAIL"
        for rule in rules.values()
    )

    warning_count = sum(
        rule["status"] == "WARNING"
        for rule in rules.values()
    )

    pass_count = sum(
        rule["status"] == "PASS"
        for rule in rules.values()
    )

    if fail_count > 0:
        risk_level = "HIGH"

    elif warning_count > 0:
        risk_level = "MEDIUM"

    else:
        risk_level = "LOW"

    return {
        "risk_level": risk_level,
        "pass_count": pass_count,
        "warning_count": warning_count,
        "fail_count": fail_count
    }
```

```python
def verify_transaction(
    ledger_df,
    history_df,
    mutation_df,
    transaction,
    people_df,
    parcels_df
):

    rules = evaluate_transaction_rules(
        ledger_df,
        history_df,
        mutation_df,
        transaction,
        people_df,
        parcels_df
    )

    risk_summary = calculate_risk_level(
        rules
    )

    return {
        "deed_id": transaction["deed_id"],
        "parcel_id": transaction["parcel_id"],
        "seller_id": transaction["seller_id"],
        "buyer_id": transaction["buyer_id"],

        "risk_level":
            risk_summary["risk_level"],

        "pass_count":
            risk_summary["pass_count"],

        "warning_count":
            risk_summary["warning_count"],

        "fail_count":
            risk_summary["fail_count"],

        "rules": rules
    }
```

# STEP 15 — Automated Test Suite

Run multiple known transaction scenarios automatically.

The test suite verifies:

1. Valid / low-risk transaction
2. Seller not owner
3. Over-transfer
4. Double-sale
5. Duplicate deed
6. Deed-mutation mismatch
7. Missing mutation record / medium risk

```python
def run_automated_tests(
    test_cases,
    ledger_df,
    history_df,
    mutation_df,
    people_df,
    parcels_df
):

    results = []

    for test in test_cases:

        report = verify_transaction(
            ledger_df,
            history_df,
            mutation_df,
            test["transaction"],
            people_df,
            parcels_df
        )

        rule_codes = [
            result["code"]
            for result in report["rules"].values()
        ]

        risk_pass = (
            report["risk_level"]
            == test["expected_risk"]
        )

        code_pass = (
            test["expected_code"]
            in rule_codes
        )

        test_pass = risk_pass and code_pass

        results.append({
            "test_name": test["name"],
            "expected_risk": test["expected_risk"],
            "actual_risk": report["risk_level"],
            "expected_code": test["expected_code"],
            "code_found": code_pass,
            "passed": test_pass
        })

    return pd.DataFrame(results)
```

# STEP 16D — Primary Rule Purity Audit

Verify that each synthetic candidate contains the intended conflict
without accidental additional critical conflicts.

For VALID records:
- No critical conflict should exist.

For each non-valid class:
- Exactly one intended critical conflict should exist.

Warnings such as missing mutation information are tracked separately
and are not treated as fraud/conflict labels.

```python
def audit_candidate_conflicts(
    transaction,
    ledger_df,
    history_df,
    mutation_df,
    people_df,
    parcels_df
):
    conflicts = []
    warnings = []

    seller_id = transaction["seller_id"]
    buyer_id = transaction["buyer_id"]
    parcel_id = transaction["parcel_id"]

    transfer_area = float(
        transaction["transferred_area_decimal"]
    )

    person_ids = set(
        people_df["person_id"]
    )

    parcel_ids = set(
        parcels_df["parcel_id"]
    )

    # ---------------------------------
    # Reference checks
    # ---------------------------------

    if seller_id not in person_ids:
        conflicts.append(
            "SELLER_NOT_FOUND"
        )

    if buyer_id not in person_ids:
        conflicts.append(
            "BUYER_NOT_FOUND"
        )

    if parcel_id not in parcel_ids:
        conflicts.append(
            "PARCEL_NOT_FOUND"
        )

        return {
            "conflicts": conflicts,
            "warnings": warnings
        }

    # ---------------------------------
    # Seller / buyer identity
    # ---------------------------------

    if seller_id == buyer_id:
        conflicts.append(
            "SELLER_BUYER_SAME"
        )

    # ---------------------------------
    # Duplicate deed
    # ---------------------------------

    duplicate_result = check_duplicate_deed(
        history_df,
        transaction["deed_id"]
    )

    if duplicate_result["duplicate"]:
        conflicts.append(
            "DUPLICATE_DEED"
        )

    # ---------------------------------
    # Double-sale
    # ---------------------------------

    double_sale_result = check_double_sale(
        history_df,
        transaction
    )

    if double_sale_result["double_sale"]:
        conflicts.append(
            "DOUBLE_SALE"
        )

    elif (
        double_sale_result["reason"]
        == "NO_PORTION_ID"
    ):
        warnings.append(
            "PORTION_ID_UNAVAILABLE"
        )

    # ---------------------------------
    # Seller ownership / area
    # ---------------------------------

    seller_area = get_owner_area(
        ledger_df,
        parcel_id,
        seller_id
    )

    if seller_area <= 0:

        conflicts.append(
            "SELLER_NOT_OWNER"
        )

    elif transfer_area <= 0:

        conflicts.append(
            "INVALID_TRANSFER_AREA"
        )

    elif transfer_area > seller_area:

        conflicts.append(
            "INSUFFICIENT_OWNERSHIP"
        )

    # ---------------------------------
    # Mutation consistency
    # ---------------------------------

    mutation_result = check_deed_mutation_match(
        mutation_df,
        transaction
    )

    if mutation_result["match"] is False:

        conflicts.append(
            "DEED_MUTATION_MISMATCH"
        )

    elif mutation_result["match"] is None:

        warnings.append(
            "NO_MUTATION_RECORD"
        )

    return {
        "conflicts": conflicts,
        "warnings": warnings
    }
```

```python
expected_conflict_map = {
    "VALID":
        None,

    "SELLER_NOT_OWNER":
        "SELLER_NOT_OWNER",

    "OVER_TRANSFER":
        "INSUFFICIENT_OWNERSHIP",

    "DOUBLE_SALE":
        "DOUBLE_SALE",

    "DUPLICATE_DEED":
        "DUPLICATE_DEED",

    "MUTATION_MISMATCH":
        "DEED_MUTATION_MISMATCH"
}
```

```python
purity_results = []

for _, row in synthetic_candidates_df.iterrows():

    transaction = {
        "deed_id":
            row["deed_id"],

        "deed_type":
            row["deed_type"],

        "execution_date":
            row["execution_date"],

        "seller_id":
            row["seller_id"],

        "buyer_id":
            row["buyer_id"],

        "parcel_id":
            row["parcel_id"],

        "portion_id":
            row["portion_id"],

        "transferred_area_decimal":
            float(
                row[
                    "transferred_area_decimal"
                ]
            ),

        "transaction_status":
            row["transaction_status"]
    }

    audit = audit_candidate_conflicts(
        transaction,
        large_reference_ledger_df,
        large_reference_history_df,
        scenario_mutation_df,
        large_people_df,
        large_parcels_df
    )

    conflicts = audit["conflicts"]
    warnings = audit["warnings"]

    expected_conflict = (
        expected_conflict_map[
            row["scenario_label"]
        ]
    )

    if expected_conflict is None:

        pure = (
            len(conflicts) == 0
        )

    else:

        pure = (
            len(conflicts) == 1
            and conflicts[0]
            == expected_conflict
        )

    purity_results.append({
        "candidate_id":
            row["candidate_id"],

        "scenario_label":
            row["scenario_label"],

        "expected_conflict":
            expected_conflict,

        "conflict_count":
            len(conflicts),

        "conflicts":
            conflicts,

        "warning_count":
            len(warnings),

        "warnings":
            warnings,

        "rule_pure":
            pure
    })
```

```python
normalization_code = r'''
import re
import unicodedata


BANGLA_DIGITS = str.maketrans(
    "০১২৩৪৫৬৭৮৯",
    "0123456789"
)


SURVEY_ALIASES = {
    "CS": "CS",
    "C S": "CS",
    "C.S": "CS",
    "C.S.": "CS",
    "সি এস": "CS",

    "SA": "SA",
    "S A": "SA",
    "S.A": "SA",
    "S.A.": "SA",
    "এস এ": "SA",

    "RS": "RS",
    "R S": "RS",
    "R.S": "RS",
    "R.S.": "RS",
    "আর এস": "RS",

    "BS": "BS",
    "B S": "BS",
    "B.S": "BS",
    "B.S.": "BS",
    "বি এস": "BS",
}


def bangla_to_english_digits(value):
    if value is None:
        return None

    return str(value).translate(
        BANGLA_DIGITS
    )


def normalize_land_number(value):
    if value is None:
        return None

    value = bangla_to_english_digits(
        value
    )

    value = unicodedata.normalize(
        "NFKC",
        value
    )

    value = value.strip()

    labels = [
        "দাগ নং",
        "দাগ নম্বর",
        "দাগ",
        "খতিয়ান নং",
        "খতিয়ান নং",
        "খতিয়ান নম্বর",
        "খতিয়ান নম্বর",
        "খতিয়ান",
        "খতিয়ান",
        "নং",
        "নম্বর",
        "NO.",
        "NO",
    ]

    for label in labels:
        value = value.replace(
            label,
            ""
        )

    value = value.strip()

    digits = re.sub(
        r"[^0-9]",
        "",
        value
    )

    if digits == "":
        return None

    return str(
        int(digits)
    )


def normalize_survey_type(value):
    if value is None:
        return None

    value = unicodedata.normalize(
        "NFKC",
        str(value)
    )

    value = " ".join(
        value.strip()
        .upper()
        .split()
    )

    return SURVEY_ALIASES.get(
        value,
        value
    )


def normalize_location(value):
    if value is None:
        return None

    value = unicodedata.normalize(
        "NFKC",
        str(value)
    )

    value = " ".join(
        value.strip()
        .split()
    )

    return value.upper()


def create_parcel_key(
    district,
    upazila,
    mouza,
    survey_type,
    khatian_no,
    dag_no
):
    district_n = normalize_location(
        district
    )

    upazila_n = normalize_location(
        upazila
    )

    mouza_n = normalize_location(
        mouza
    )

    survey_n = normalize_survey_type(
        survey_type
    )

    khatian_n = normalize_land_number(
        khatian_no
    )

    dag_n = normalize_land_number(
        dag_no
    )

    components = [
        district_n,
        upazila_n,
        mouza_n,
        survey_n,
        khatian_n,
        dag_n,
    ]

    if any(
        value is None
        for value in components
    ):
        return None

    return "|".join(
        components
    )
'''

normalization_file = (
    BHUMIGUARD_SRC
    / "normalization.py"
)

normalization_file.write_text(
    normalization_code.strip()
    + "\n",
    encoding="utf-8"
)

print(
    "normalization.py created:",
    normalization_file.exists()
)
```

# STEP 17B — Reusable Ownership Ledger Module

Move ownership and transaction-processing logic into:

src/bhumiguard/ledger.py

Responsibilities:

- current owner-area lookup
- ownership ID generation
- ledger-integrity validation
- safe transfer processing
- preservation of the original ledger when a transaction is rejected

```python
ledger_code = r'''
import pandas as pd


def get_owner_area(
    ledger_df,
    parcel_id,
    owner_id
):
    rows = ledger_df[
        (ledger_df["parcel_id"] == parcel_id) &
        (ledger_df["owner_id"] == owner_id) &
        (ledger_df["status"] == "ACTIVE")
    ]

    if rows.empty:
        return 0.0

    return float(
        rows["owned_area_decimal"].sum()
    )


def next_ownership_id(ledger_df):
    numbers = []

    for value in ledger_df["ownership_id"]:
        value = str(value)

        if value.startswith("OWN"):

            number = value[3:]

            if number.isdigit():
                numbers.append(
                    int(number)
                )

    next_number = (
        max(numbers, default=0) + 1
    )

    return f"OWN{next_number:03d}"


def check_ledger_integrity(
    ledger_df,
    parcels_df,
    tolerance=1e-9
):
    required_ledger_columns = {
        "ownership_id",
        "parcel_id",
        "owner_id",
        "owned_area_decimal",
        "status",
    }

    required_parcel_columns = {
        "parcel_id",
        "total_area_decimal",
    }

    if not required_ledger_columns.issubset(
        ledger_df.columns
    ):
        return False

    if not required_parcel_columns.issubset(
        parcels_df.columns
    ):
        return False

    # No negative ownership
    if (
        ledger_df[
            "owned_area_decimal"
        ] < 0
    ).any():
        return False

    # Every ownership row must reference a real parcel
    if (
        ~ledger_df[
            "parcel_id"
        ].isin(
            parcels_df[
                "parcel_id"
            ]
        )
    ).any():
        return False

    active = ledger_df[
        ledger_df["status"] == "ACTIVE"
    ].copy()

    active_totals = (
        active
        .groupby(
            "parcel_id",
            as_index=False
        )[
            "owned_area_decimal"
        ]
        .sum()
    )

    parcel_totals = parcels_df[
        [
            "parcel_id",
            "total_area_decimal"
        ]
    ].copy()

    comparison = parcel_totals.merge(
        active_totals,
        on="parcel_id",
        how="left"
    )

    comparison[
        "owned_area_decimal"
    ] = comparison[
        "owned_area_decimal"
    ].fillna(0.0)

    difference = (
        comparison[
            "owned_area_decimal"
        ]
        -
        comparison[
            "total_area_decimal"
        ]
    ).abs()

    return bool(
        (difference <= tolerance).all()
    )


def process_transfer(
    ledger_df,
    transaction,
    people_df,
    parcels_df
):
    original_ledger = (
        ledger_df.copy(deep=True)
    )

    updated_ledger = (
        ledger_df.copy(deep=True)
    )

    seller_id = transaction["seller_id"]
    buyer_id = transaction["buyer_id"]
    parcel_id = transaction["parcel_id"]

    transfer_area = float(
        transaction[
            "transferred_area_decimal"
        ]
    )

    person_ids = set(
        people_df["person_id"]
    )

    parcel_ids = set(
        parcels_df["parcel_id"]
    )

    # ---------------------------------
    # Basic validation
    # ---------------------------------

    if seller_id not in person_ids:

        return original_ledger, {
            "valid": False,
            "reason": "SELLER_NOT_FOUND"
        }

    if buyer_id not in person_ids:

        return original_ledger, {
            "valid": False,
            "reason": "BUYER_NOT_FOUND"
        }

    if parcel_id not in parcel_ids:

        return original_ledger, {
            "valid": False,
            "reason": "PARCEL_NOT_FOUND"
        }

    if seller_id == buyer_id:

        return original_ledger, {
            "valid": False,
            "reason": "SELLER_BUYER_SAME"
        }

    if transfer_area <= 0:

        return original_ledger, {
            "valid": False,
            "reason": "INVALID_TRANSFER_AREA"
        }

    # ---------------------------------
    # Seller ownership
    # ---------------------------------

    seller_available = get_owner_area(
        updated_ledger,
        parcel_id,
        seller_id
    )

    if seller_available <= 0:

        return original_ledger, {
            "valid": False,
            "reason": "SELLER_NOT_OWNER",
            "seller_available": 0.0
        }

    if transfer_area > seller_available:

        return original_ledger, {
            "valid": False,
            "reason":
                "INSUFFICIENT_OWNERSHIP",

            "seller_available":
                seller_available,

            "requested_area":
                transfer_area
        }

    # ---------------------------------
    # Deduct seller ownership
    # ---------------------------------

    seller_indices = updated_ledger[
        (updated_ledger["parcel_id"] == parcel_id) &
        (updated_ledger["owner_id"] == seller_id) &
        (updated_ledger["status"] == "ACTIVE")
    ].index.tolist()

    remaining_to_deduct = transfer_area

    for index in seller_indices:

        row_area = float(
            updated_ledger.loc[
                index,
                "owned_area_decimal"
            ]
        )

        deduction = min(
            row_area,
            remaining_to_deduct
        )

        new_area = (
            row_area - deduction
        )

        updated_ledger.loc[
            index,
            "owned_area_decimal"
        ] = new_area

        if abs(new_area) <= 1e-9:

            updated_ledger.loc[
                index,
                "owned_area_decimal"
            ] = 0.0

            updated_ledger.loc[
                index,
                "status"
            ] = "INACTIVE"

        remaining_to_deduct -= (
            deduction
        )

        if remaining_to_deduct <= 1e-9:
            break

    # ---------------------------------
    # Add buyer ownership
    # ---------------------------------

    buyer_indices = updated_ledger[
        (updated_ledger["parcel_id"] == parcel_id) &
        (updated_ledger["owner_id"] == buyer_id) &
        (updated_ledger["status"] == "ACTIVE")
    ].index.tolist()

    if buyer_indices:

        buyer_index = buyer_indices[0]

        updated_ledger.loc[
            buyer_index,
            "owned_area_decimal"
        ] = (
            float(
                updated_ledger.loc[
                    buyer_index,
                    "owned_area_decimal"
                ]
            )
            + transfer_area
        )

    else:

        new_row = {
            "ownership_id":
                next_ownership_id(
                    updated_ledger
                ),

            "parcel_id":
                parcel_id,

            "owner_id":
                buyer_id,

            "owned_area_decimal":
                transfer_area,

            "valid_from":
                transaction[
                    "execution_date"
                ],

            "status":
                "ACTIVE"
        }

        updated_ledger = pd.concat(
            [
                updated_ledger,
                pd.DataFrame(
                    [new_row]
                )
            ],
            ignore_index=True
        )

    # ---------------------------------
    # Final integrity protection
    # ---------------------------------

    if not check_ledger_integrity(
        updated_ledger,
        parcels_df
    ):

        return original_ledger, {
            "valid": False,
            "reason":
                "LEDGER_INTEGRITY_FAILURE"
        }

    return updated_ledger, {
        "valid": True,
        "reason": "VALID_TRANSFER",

        "seller_available_before":
            seller_available,

        "transferred_area":
            transfer_area
    }
'''

ledger_file = (
    BHUMIGUARD_SRC
    / "ledger.py"
)

ledger_file.write_text(
    ledger_code.strip() + "\n",
    encoding="utf-8"
)

print(
    "ledger.py created:",
    ledger_file.exists()
)
```

# STEP 17C — Reusable Validation Module

Move transaction-history and cross-record validation logic into:

src/bhumiguard/validation.py

Responsibilities:

- duplicate deed detection
- repeated portion / double-sale detection
- deed-mutation consistency checking
- combined transaction validation
- preserving ledger/history when a transaction is rejected

```python
validation_code = r'''
import pandas as pd

from .ledger import process_transfer


def check_duplicate_deed(
    history_df,
    deed_id
):
    matches = history_df[
        history_df["deed_id"] == deed_id
    ]

    if matches.empty:
        return {
            "duplicate": False,
            "reason": "DEED_ID_AVAILABLE"
        }

    return {
        "duplicate": True,
        "reason": "DUPLICATE_DEED",
        "existing_records": len(matches)
    }


def check_double_sale(
    history_df,
    transaction
):
    portion_id = transaction.get(
        "portion_id"
    )

    if portion_id is None:
        return {
            "double_sale": False,
            "reason": "NO_PORTION_ID"
        }

    required_columns = {
        "parcel_id",
        "portion_id",
        "transaction_status",
        "deed_id",
        "buyer_id",
    }

    if not required_columns.issubset(
        history_df.columns
    ):
        return {
            "double_sale": False,
            "reason": "HISTORY_SCHEMA_INCOMPLETE"
        }

    previous_sales = history_df[
        (
            history_df["parcel_id"]
            == transaction["parcel_id"]
        )
        &
        (
            history_df["portion_id"]
            == portion_id
        )
        &
        (
            history_df[
                "transaction_status"
            ]
            == "VALID"
        )
    ]

    if previous_sales.empty:
        return {
            "double_sale": False,
            "reason": "NO_PREVIOUS_SALE"
        }

    return {
        "double_sale": True,
        "reason": "DOUBLE_SALE",

        "previous_deeds":
            previous_sales[
                "deed_id"
            ].tolist(),

        "previous_buyers":
            previous_sales[
                "buyer_id"
            ].tolist(),
    }


def check_deed_mutation_match(
    mutation_df,
    transaction
):
    required_columns = {
        "parcel_id",
        "portion_id",
        "mutation_owner_id",
        "status",
    }

    if not required_columns.issubset(
        mutation_df.columns
    ):
        return {
            "match": None,
            "reason":
                "MUTATION_SCHEMA_INCOMPLETE"
        }

    records = mutation_df[
        (
            mutation_df["parcel_id"]
            == transaction["parcel_id"]
        )
        &
        (
            mutation_df["portion_id"]
            == transaction.get(
                "portion_id"
            )
        )
        &
        (
            mutation_df["status"]
            == "ACTIVE"
        )
    ]

    if records.empty:
        return {
            "match": None,
            "reason": "NO_MUTATION_RECORD"
        }

    mutation_owners = set(
        records[
            "mutation_owner_id"
        ]
    )

    deed_seller = transaction[
        "seller_id"
    ]

    if deed_seller in mutation_owners:
        return {
            "match": True,
            "reason": "MUTATION_MATCH",
            "mutation_owners":
                sorted(
                    mutation_owners
                ),
        }

    return {
        "match": False,
        "reason":
            "DEED_MUTATION_MISMATCH",

        "deed_seller":
            deed_seller,

        "mutation_owners":
            sorted(
                mutation_owners
            ),
    }


def process_transfer_full_checks(
    ledger_df,
    history_df,
    mutation_df,
    transaction,
    people_df,
    parcels_df
):
    original_ledger = (
        ledger_df.copy(deep=True)
    )

    original_history = (
        history_df.copy(deep=True)
    )

    # ---------------------------------
    # 1. Duplicate deed
    # ---------------------------------

    duplicate_result = (
        check_duplicate_deed(
            history_df,
            transaction["deed_id"]
        )
    )

    if duplicate_result["duplicate"]:

        return (
            original_ledger,
            original_history,
            {
                "valid": False,
                "reason": "DUPLICATE_DEED",
                "existing_records":
                    duplicate_result[
                        "existing_records"
                    ],
            }
        )

    # ---------------------------------
    # 2. Double-sale
    # ---------------------------------

    double_sale_result = (
        check_double_sale(
            history_df,
            transaction
        )
    )

    if double_sale_result[
        "double_sale"
    ]:

        return (
            original_ledger,
            original_history,
            {
                "valid": False,
                "reason": "DOUBLE_SALE",

                "previous_deeds":
                    double_sale_result[
                        "previous_deeds"
                    ],

                "previous_buyers":
                    double_sale_result[
                        "previous_buyers"
                    ],
            }
        )

    # ---------------------------------
    # 3. Mutation consistency
    # ---------------------------------

    mutation_result = (
        check_deed_mutation_match(
            mutation_df,
            transaction
        )
    )

    if mutation_result["match"] is False:

        return (
            original_ledger,
            original_history,
            {
                "valid": False,
                "reason":
                    "DEED_MUTATION_MISMATCH",

                "deed_seller":
                    mutation_result[
                        "deed_seller"
                    ],

                "mutation_owners":
                    mutation_result[
                        "mutation_owners"
                    ],
            }
        )

    # ---------------------------------
    # 4. Ownership / transfer validation
    # ---------------------------------

    (
        updated_ledger,
        transfer_result
    ) = process_transfer(
        ledger_df,
        transaction,
        people_df,
        parcels_df
    )

    if not transfer_result["valid"]:

        return (
            original_ledger,
            original_history,
            transfer_result
        )

    # ---------------------------------
    # 5. Store accepted transaction
    # ---------------------------------

    accepted_transaction = (
        transaction.copy()
    )

    accepted_transaction[
        "transaction_status"
    ] = "VALID"

    updated_history = pd.concat(
        [
            original_history,
            pd.DataFrame(
                [accepted_transaction]
            )
        ],
        ignore_index=True
    )

    return (
        updated_ledger,
        updated_history,
        {
            "valid": True,
            "reason": "VALID_TRANSFER",

            "mutation_status":
                mutation_result[
                    "reason"
                ],
        }
    )
'''

validation_file = (
    BHUMIGUARD_SRC
    / "validation.py"
)

validation_file.write_text(
    validation_code.strip() + "\n",
    encoding="utf-8"
)

print(
    "validation.py created:",
    validation_file.exists()
)
```

# STEP 17D — Explainable Risk Engine Module

Move the explainable pre-registration screening logic into:

src/bhumiguard/risk.py

Responsibilities:

- independently evaluate verification rules
- produce PASS / WARNING / FAIL results
- calculate LOW / MEDIUM / HIGH screening risk
- generate explainable rule messages
- never modify ownership or transaction history

```python
risk_code = r'''
from .ledger import get_owner_area

from .validation import (
    check_duplicate_deed,
    check_double_sale,
    check_deed_mutation_match,
)


def evaluate_transaction_rules(
    ledger_df,
    history_df,
    mutation_df,
    transaction,
    people_df,
    parcels_df
):
    rules = {}

    seller_id = transaction["seller_id"]
    buyer_id = transaction["buyer_id"]
    parcel_id = transaction["parcel_id"]
    deed_id = transaction["deed_id"]

    transfer_area = float(
        transaction[
            "transferred_area_decimal"
        ]
    )

    person_ids = set(
        people_df["person_id"]
    )

    parcel_ids = set(
        parcels_df["parcel_id"]
    )

    # ---------------------------------
    # Rule 1: Deed ID
    # ---------------------------------

    duplicate_result = (
        check_duplicate_deed(
            history_df,
            deed_id
        )
    )

    if duplicate_result["duplicate"]:

        rules["deed_id_check"] = {
            "status": "FAIL",
            "code": "DUPLICATE_DEED",
            "message": (
                f"Deed ID {deed_id} "
                "already exists in the "
                "accepted transaction history."
            )
        }

    else:

        rules["deed_id_check"] = {
            "status": "PASS",
            "code": "DEED_ID_AVAILABLE",
            "message":
                "Deed ID is not duplicated."
        }

    # ---------------------------------
    # Rule 2: Party references
    # ---------------------------------

    if seller_id not in person_ids:

        rules["seller_reference_check"] = {
            "status": "FAIL",
            "code": "SELLER_NOT_FOUND",
            "message":
                "Seller was not found in the registry."
        }

    else:

        rules["seller_reference_check"] = {
            "status": "PASS",
            "code": "SELLER_FOUND",
            "message":
                "Seller exists in the registry."
        }

    if buyer_id not in person_ids:

        rules["buyer_reference_check"] = {
            "status": "FAIL",
            "code": "BUYER_NOT_FOUND",
            "message":
                "Buyer was not found in the registry."
        }

    elif seller_id == buyer_id:

        rules["buyer_reference_check"] = {
            "status": "FAIL",
            "code": "SELLER_BUYER_SAME",
            "message":
                "Seller and buyer cannot be the same party."
        }

    else:

        rules["buyer_reference_check"] = {
            "status": "PASS",
            "code": "BUYER_FOUND",
            "message":
                "Buyer exists and differs from the seller."
        }

    # ---------------------------------
    # Rule 3: Parcel reference
    # ---------------------------------

    parcel_exists = (
        parcel_id in parcel_ids
    )

    if parcel_exists:

        rules["parcel_check"] = {
            "status": "PASS",
            "code": "PARCEL_FOUND",
            "message":
                "Parcel exists in the registry."
        }

    else:

        rules["parcel_check"] = {
            "status": "FAIL",
            "code": "PARCEL_NOT_FOUND",
            "message":
                "Parcel was not found in the registry."
        }

    # ---------------------------------
    # Rule 4: Seller ownership
    # ---------------------------------

    seller_area = get_owner_area(
        ledger_df,
        parcel_id,
        seller_id
    )

    if seller_area > 0:

        rules["seller_ownership_check"] = {
            "status": "PASS",
            "code": "SELLER_IS_OWNER",
            "message": (
                f"Seller currently owns "
                f"{seller_area} decimal "
                "in the parcel."
            )
        }

    else:

        rules["seller_ownership_check"] = {
            "status": "FAIL",
            "code": "SELLER_NOT_OWNER",
            "message":
                "Seller has no active ownership "
                "in this parcel."
        }

    # ---------------------------------
    # Rule 5: Transfer area
    # ---------------------------------

    if transfer_area <= 0:

        rules["area_check"] = {
            "status": "FAIL",
            "code":
                "INVALID_TRANSFER_AREA",
            "message":
                "Transfer area must be positive."
        }

    elif (
        seller_area > 0
        and transfer_area > seller_area
    ):

        rules["area_check"] = {
            "status": "FAIL",
            "code":
                "INSUFFICIENT_OWNERSHIP",

            "message": (
                f"Requested transfer is "
                f"{transfer_area} decimal, "
                f"but the seller currently "
                f"owns {seller_area} decimal."
            )
        }

    elif seller_area <= 0:

        rules["area_check"] = {
            "status": "WARNING",
            "code":
                "AREA_CHECK_NOT_APPLICABLE",

            "message":
                "Transfer-area sufficiency cannot "
                "be confirmed because seller "
                "ownership is unavailable."
        }

    else:

        rules["area_check"] = {
            "status": "PASS",
            "code": "AREA_AVAILABLE",
            "message": (
                f"Requested transfer of "
                f"{transfer_area} decimal "
                "is within seller ownership."
            )
        }

    # ---------------------------------
    # Rule 6: Double-sale
    # ---------------------------------

    double_sale_result = (
        check_double_sale(
            history_df,
            transaction
        )
    )

    if double_sale_result[
        "double_sale"
    ]:

        rules["double_sale_check"] = {
            "status": "FAIL",
            "code": "DOUBLE_SALE",
            "message":
                "The same identified land portion "
                "appears in a previous valid sale."
        }

    elif (
        double_sale_result["reason"]
        == "NO_PORTION_ID"
    ):

        rules["double_sale_check"] = {
            "status": "WARNING",
            "code":
                "PORTION_ID_UNAVAILABLE",
            "message":
                "Double-sale verification is "
                "incomplete because no portion "
                "identifier is available."
        }

    elif (
        double_sale_result["reason"]
        == "HISTORY_SCHEMA_INCOMPLETE"
    ):

        rules["double_sale_check"] = {
            "status": "WARNING",
            "code":
                "HISTORY_SCHEMA_INCOMPLETE",
            "message":
                "Transaction history does not contain "
                "all fields required for double-sale "
                "verification."
        }

    else:

        rules["double_sale_check"] = {
            "status": "PASS",
            "code": "NO_PREVIOUS_SALE",
            "message":
                "No previous valid sale was found "
                "for this identified portion."
        }

    # ---------------------------------
    # Rule 7: Mutation consistency
    # ---------------------------------

    mutation_result = (
        check_deed_mutation_match(
            mutation_df,
            transaction
        )
    )

    if mutation_result["match"] is True:

        rules["mutation_check"] = {
            "status": "PASS",
            "code": "MUTATION_MATCH",
            "message":
                "Deed seller matches the active "
                "mutation-side owner."
        }

    elif mutation_result["match"] is False:

        mutation_owners = ", ".join(
            mutation_result[
                "mutation_owners"
            ]
        )

        rules["mutation_check"] = {
            "status": "FAIL",
            "code":
                "DEED_MUTATION_MISMATCH",

            "message": (
                f"Deed seller is {seller_id}, "
                f"while the mutation-side record "
                f"lists {mutation_owners}."
            )
        }

    elif (
        mutation_result["reason"]
        == "MUTATION_SCHEMA_INCOMPLETE"
    ):

        rules["mutation_check"] = {
            "status": "WARNING",
            "code":
                "MUTATION_SCHEMA_INCOMPLETE",

            "message":
                "Mutation data does not contain "
                "all fields required for verification."
        }

    else:

        rules["mutation_check"] = {
            "status": "WARNING",
            "code": "NO_MUTATION_RECORD",
            "message":
                "No matching mutation record is "
                "available for this parcel portion."
        }

    return rules


def calculate_risk_level(rules):
    pass_count = sum(
        result["status"] == "PASS"
        for result in rules.values()
    )

    warning_count = sum(
        result["status"] == "WARNING"
        for result in rules.values()
    )

    fail_count = sum(
        result["status"] == "FAIL"
        for result in rules.values()
    )

    if fail_count > 0:
        risk_level = "HIGH"

    elif warning_count > 0:
        risk_level = "MEDIUM"

    else:
        risk_level = "LOW"

    return {
        "risk_level":
            risk_level,

        "pass_count":
            pass_count,

        "warning_count":
            warning_count,

        "fail_count":
            fail_count,
    }


def verify_transaction(
    ledger_df,
    history_df,
    mutation_df,
    transaction,
    people_df,
    parcels_df
):
    rules = evaluate_transaction_rules(
        ledger_df,
        history_df,
        mutation_df,
        transaction,
        people_df,
        parcels_df
    )

    summary = calculate_risk_level(
        rules
    )

    return {
        "deed_id":
            transaction["deed_id"],

        "parcel_id":
            transaction["parcel_id"],

        "seller_id":
            transaction["seller_id"],

        "buyer_id":
            transaction["buyer_id"],

        "risk_level":
            summary["risk_level"],

        "pass_count":
            summary["pass_count"],

        "warning_count":
            summary["warning_count"],

        "fail_count":
            summary["fail_count"],

        "rules":
            rules,

        "disclaimer":
            (
                "Screening result only. "
                "It does not establish legal title "
                "or prove fraud."
            ),
    }


def format_risk_report(report):
    lines = []

    lines.append(
        "BHUMIGUARD BD — PRE-REGISTRATION CHECK"
    )

    lines.append(
        "=" * 50
    )

    lines.append(
        f"Deed: {report['deed_id']}"
    )

    lines.append(
        f"Parcel: {report['parcel_id']}"
    )

    lines.append(
        f"Seller: {report['seller_id']}"
    )

    lines.append(
        f"Buyer: {report['buyer_id']}"
    )

    lines.append("")
    lines.append("RULE CHECKS")
    lines.append("-" * 50)

    for result in report[
        "rules"
    ].values():

        lines.append(
            f"{result['status']:7} | "
            f"{result['code']}"
        )

        lines.append(
            f"          {result['message']}"
        )

    lines.append("")
    lines.append("SUMMARY")
    lines.append("-" * 50)

    lines.append(
        f"PASS: {report['pass_count']}"
    )

    lines.append(
        f"WARNING: {report['warning_count']}"
    )

    lines.append(
        f"FAIL: {report['fail_count']}"
    )

    lines.append(
        f"OVERALL RISK: "
        f"{report['risk_level']}"
    )

    lines.append("")
    lines.append(
        report["disclaimer"]
    )

    return "\n".join(lines)
'''

risk_file = (
    BHUMIGUARD_SRC
    / "risk.py"
)

risk_file.write_text(
    risk_code.strip() + "\n",
    encoding="utf-8"
)

print(
    "risk.py created:",
    risk_file.exists()
)
```
