# BhumiGuard BD — Step 1: Data Setup

This notebook verifies the source-data setup for **LandTech 2026 — P3: Catch Fraud & Double-Selling**.

Public/sample documents are used only as source/reference material. Ownership and fraud ground truth will be created later using privacy-safe synthetic records.

## 1. Locate the project root

This cell finds the `BhumiGuard-BD` project folder whether the notebook is opened from VS Code or Jupyter.

```python
from pathlib import Path

cwd = Path.cwd().resolve()

if cwd.name == 'notebooks':
    PROJECT_ROOT = cwd.parent
elif (cwd / 'notebooks').exists() and (cwd / 'data').exists():
    PROJECT_ROOT = cwd
else:
    raise FileNotFoundError('Open this notebook from inside the BhumiGuard-BD project folder.')

print('Project root:', PROJECT_ROOT)
```

## 2. Verify required folders

No data is modified in this step.

```python
required_folders = [
    'data/raw/handwritten_dolil',
    'data/raw/deed_samples',
    'data/raw/dlrs_forms',
    'data/synthetic',
    'data/processed',
    'src',
    'tests',
    'app'
]

for folder in required_folders:
    path = PROJECT_ROOT / folder
    print(f"{'OK' if path.exists() else 'MISSING':7}  {folder}")
```

## 3. Check downloaded source files

Place your downloaded deed PDFs inside `data/raw/deed_samples/` and extract the Mendeley handwritten Dolil dataset inside `data/raw/handwritten_dolil/` before running this cell.

```python
deed_dir = PROJECT_ROOT / 'data/raw/deed_samples'
handwritten_dir = PROJECT_ROOT / 'data/raw/handwritten_dolil'
dlrs_dir = PROJECT_ROOT / 'data/raw/dlrs_forms'

deed_files = [p for p in deed_dir.rglob('*') if p.is_file() and p.name != '.gitkeep']
handwritten_files = [p for p in handwritten_dir.rglob('*') if p.is_file() and p.name != '.gitkeep']
dlrs_files = [p for p in dlrs_dir.rglob('*') if p.is_file() and p.name != '.gitkeep']

print('Deed sample files:', len(deed_files))
print('Handwritten files:', len(handwritten_files))
print('DLRS files:', len(dlrs_files))

print('\nDeed samples:')
for file in deed_files:
    print('-', file.relative_to(deed_dir))
```

## 4. Count image files by extension

This is only an audit. It does not open, edit, rename, resize, or convert any source image.

```python
from collections import Counter

image_exts = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff', '.webp'}
image_files = [p for p in handwritten_files if p.suffix.lower() in image_exts]
ext_counts = Counter(p.suffix.lower() for p in image_files)

print('Handwritten image files:', len(image_files))
print('Image extension counts:')
for ext, count in sorted(ext_counts.items()):
    print(f'- {ext}: {count}')

other_files = [p for p in handwritten_files if p.suffix.lower() not in image_exts]
print('Non-image files inside handwritten folder:', len(other_files))
```
