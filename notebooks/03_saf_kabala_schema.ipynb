# BhumiGuard BD — STEP 3: Saf Kabala Field Schema

This notebook defines the minimum structured fields required from a
Saf Kabala deed for P3: Fraud & Double-Selling detection.

The schema is derived from the collected Saf Kabala sample document.

No OCR or raw-data modification is performed in this step.

```python
core_fields = [
    "deed_id",
    "deed_type",
    "registration_office",
    "execution_date",

    "seller_id",
    "seller_name",

    "buyer_id",
    "buyer_name",

    "district",
    "upazila",
    "mouza",

    "survey_type",
    "khatian_no",
    "dag_no",

    "total_property_area",
    "transferred_area",
    "seller_ownership_share",

    "previous_deed_reference",
    "power_of_attorney_reference"
]

print("Core P3 fields:", len(core_fields))

for i, field in enumerate(core_fields, start=1):
    print(f"{i:02d}. {field}")
```

```python
field_groups = {
    "deed": [
        "deed_id",
        "deed_type",
        "registration_office",
        "execution_date"
    ],

    "parties": [
        "seller_id",
        "seller_name",
        "buyer_id",
        "buyer_name"
    ],

    "parcel": [
        "district",
        "upazila",
        "mouza",
        "survey_type",
        "khatian_no",
        "dag_no"
    ],

    "ownership": [
        "total_property_area",
        "transferred_area",
        "seller_ownership_share"
    ],

    "history": [
        "previous_deed_reference",
        "power_of_attorney_reference"
    ]
}

for group, fields in field_groups.items():
    print(f"\n{group.upper()} ({len(fields)})")
    for field in fields:
        print("-", field)
```

```python
parcel_identity_fields = [
    "district",
    "upazila",
    "mouza",
    "survey_type",
    "khatian_no",
    "dag_no"
]

print("Parcel identity fields:")

for field in parcel_identity_fields:
    print("-", field)
```

```python
fraud_detection_fields = [
    "seller_id",
    "buyer_id",
    "parcel_key",
    "transferred_area",
    "seller_ownership_share",
    "execution_date",
    "deed_id",
    "previous_deed_reference"
]

print("Fraud-detection fields:")

for field in fraud_detection_fields:
    print("-", field)
```

```python
print("STEP 3 SCHEMA SUMMARY")
print("---------------------")

print("Core fields:", len(core_fields))
print("Field groups:", len(field_groups))
print("Parcel identity fields:", len(parcel_identity_fields))
print("Fraud-detection fields:", len(fraud_detection_fields))

print("\nPrimary supported deed type:")
print("SAF_KABALA")

print("\nSchema status: READY")
```
